```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
```

```python
data=pd.read_csv('../Datos en bruto/Google Trends/Mosquito IT/multiTimeline.csv',skiprows=2,sep=",")
data.head()
```

```python
for i in range(1,17):
    df=pd.read_csv('../Datos en bruto/Google Trends/Mosquito IT/multiTimeline ('+str(i)+').csv',skiprows=2,sep=",")
    data=pd.concat([data,df])
```

```python
data=data.drop_duplicates(subset='Semana',keep='last')
```

```python
data['Semana'] = pd.to_datetime(data['Semana'])
data.set_index('Semana',inplace=True)
data=data.sort_index()
data=data.rename({'mosquito: (Italia)':'mosquito_it_trends'},axis=1)
data
```

```python
dat=data.resample('MS').sum()
dat
```

```python
#Visualización de datos mensuales
plt.figure(figsize=(13,5))
sns.lineplot(data=dat,x=dat.index,y=dat.mosquito_it_trends)
plt.xticks(rotation=90)
plt.show()
```

```python
#Exportamos los datos relativos a las búsquedas de mosquito en Italia
data.to_csv('mosquito_it_trends.csv',index=True,encoding='utf-8')
```
