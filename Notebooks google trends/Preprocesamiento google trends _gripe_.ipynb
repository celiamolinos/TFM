```python
import pandas as pd
from datetime import datetime
```

```python
data=pd.read_csv('../Datos en bruto/Google Trends/Gripe/multiTimeline.csv',skiprows=2,sep=",")
data.head()
```

```python
for i in range(1,13):
    df=pd.read_csv('../Datos en bruto/Google Trends/Gripe/multiTimeline ('+str(i)+').csv',skiprows=2,sep=",")
    data=pd.concat([data,df])
```

```python
data=data.drop_duplicates(subset='Semana',keep='last')
```

```python
data.set_index('Semana',inplace=True)
data=data.sort_index()
data=data.rename({'gripe: (España)':'gripe_trends'},axis=1)
data
```

```python
#Exportamos los datos relativos a las busquedas de gripe en España
data.to_csv('flu_trends.csv',index=True,encoding='utf-8')
```
