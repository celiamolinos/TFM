```python
import datetime
def primer_domingo_tr(anio,mes):
    fecha = datetime.date(anio, mes, 1)  # 1 del mes dado del año dado
    dias_hasta_domingo = (6 - fecha.weekday()) % 7  # Calcular días hasta el próximo domingo
    return fecha + datetime.timedelta(days=dias_hasta_domingo)
```

```python
ine08['Fecha'] = pd.to_datetime(ine08.apply(lambda row: primer_domingo_tr(row['Año'], row['mes']), axis=1))

ine=ine08.iloc[:,[4,6,10]]
ine.set_index('Fecha', inplace=True)
ine
```

### Tratamiento de los valores nulos

```python
tot = tot.dropna(subset=['COUNTRY_CODE'])
tot
```

```python
import numpy as np
import pandas as pd

def interpolate_and_extrapolate(series):
    values = series.values.copy()
    n = len(values)
    
    # Índices no nulos
    non_nan_idx = np.where(~np.isnan(values))[0]
    
    if len(non_nan_idx) < 2:
        raise ValueError("Se necesitan al menos dos valores no nulos para interpolar y extrapolar.")
    
    # Interpolación interna 
    values = np.interp(x=np.arange(n), xp=non_nan_idx, fp=values[non_nan_idx])
    
    # Calculamos el primer y último step
    first_step = (values[non_nan_idx[1]] - values[non_nan_idx[0]]) / (non_nan_idx[1] - non_nan_idx[0])
    last_step = (values[non_nan_idx[-1]] - values[non_nan_idx[-2]]) / (non_nan_idx[-1] - non_nan_idx[-2])
    
    # Extrapolación a la izquierda (inicio)
    for i in range(0, non_nan_idx[0]):
        values[i] = values[non_nan_idx[0]] - first_step * (non_nan_idx[0] - i)
    
    # Extrapolación a la derecha (final)
    for i in range(non_nan_idx[-1] + 1, n):
        values[i] = values[non_nan_idx[-1]] + last_step * (i - non_nan_idx[-1])
        
    return pd.Series(values, index=series.index)

tot.value=interpolate_and_extrapolate(tot.value)
```

### Cálculo de la ventana deslizante

```python
def one_step_forecast(df, window):
    d = df.values
    x = []
    n = len(df)
    idx = df.index[:-window]
    for start in range(n-window):
        end = start + window
        x.append(d[start:end])
    cols = [f'x_{i}' for i in range(1, window+1)]
    x = np.array(x).reshape(n-window, -1)
    y = df.iloc[window:].values
    df_xs = pd.DataFrame(x, columns=cols, index=idx)
    df_y = pd.DataFrame(y.reshape(-1), columns=['y'], index=idx)
    return pd.concat([df_y,df_xs], axis=1).dropna()
```

```python
df_inf=one_step_forecast(df.iloc[:,0],3) #ventana deslizante de infectados
df_temp=one_step_forecast(df.iloc[:,1],3).iloc[:,1:].rename({'x_1':'x_1_temp','x_2':'x_2_temp','x_3':'x_3_temp'},axis=1) #ventana deslizante de temperatura
df_hum=one_step_forecast(df.iloc[:,2],3).iloc[:,1:].rename({'x_1':'x_1_hum','x_2':'x_2_hum','x_3':'x_3_hum'},axis=1) #ventana deslizante de humedad
df_rain=one_step_forecast(df.iloc[:,3],3).iloc[:,1:].rename({'x_1':'x_1_rain','x_2':'x_2_rain','x_3':'x_3_rain'},axis=1) #ventana deslizante de precipitaciones
df_wind=one_step_forecast(df.iloc[:,4],3).iloc[:,1:].rename({'x_1':'x_1_wind','x_2':'x_2_wind','x_3':'x_3_wind'},axis=1) #ventana deslizante de fuerza del viento
df_pop=one_step_forecast(df.iloc[:,5],3).iloc[:,1:].rename({'x_1':'x_1_pop','x_2':'x_2_pop','x_3':'x_3_pop'},axis=1) #poblacion
```

```python
df=pd.concat([df_inf,df_temp,df_hum,df_rain,df_wind,df_pop],axis=1)
df
```

### Normalización y separación en test y train 

```python
def split_data(df,test_split=0.1):
    n=int(len(df)*test_split)
    train,test=df[:-n],df[-n:]
    return train, test
```

```python
train,test=split_data(df)
```

```python
from sklearn.preprocessing import MinMaxScaler
norm=MinMaxScaler()
train_norm=pd.DataFrame(norm.fit_transform(train),index=train.index,columns=train.columns)
test_norm=pd.DataFrame(norm.transform(test),index=test.index,columns=test.columns)
```

```python
xtrainN=train_norm.iloc[:,1:]
xtestN=test_norm.iloc[:,1:]
ytrainN=train_norm.iloc[:,0]
ytestN=test_norm.iloc[:,0]
```

### Regresión lineal

```python
from sklearn import linear_model
regr=linear_model.LinearRegression()
regr.fit(xtrainN,ytrainN)
ypred=regr.predict(xtestN)
Y_rl=pd.DataFrame(ytestN,index=ytestN.index).rename({'y':'ytest'},axis=1)
Y_rl['ypred']=pd.DataFrame(ypred,index=ytestN.index)
Y_rl
```

```python
from sklearn.metrics import mean_squared_error,r2_score
mse=mean_squared_error(ytestN, ypred)
r2=r2_score(ytestN, ypred)
print(mse)
print(r2)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(13,5))
sns.lineplot(data=Y_rl,x=Y_rl.index,y=Y_rl.ytest,label='ytest')
sns.lineplot(data=Y_rl,x=Y_rl.index,y=Y_rl.ypred,label='ypred')
plt.legend()
plt.xticks(rotation=90)
plt.text(datetime(2024, 3, 1), 0.95*max(Y_rl.ytest), 'La precisión de la predicción es de '+str(r2_score(Y_rl.ytest, Y_rl.ypred)), fontsize=10)
plt.show()
```
