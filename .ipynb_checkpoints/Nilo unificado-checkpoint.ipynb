### Datos de la enfermedad del Nilo en Italia

```python
import pandas as pd
data=pd.read_csv("./surveillance atlas (Hepatitis A, Nilo, Malaria)/ECDC_surveillance_data_West_Nile_virus_infection.csv", sep=",")
dataIT=data[(data.Indicator=='Reported cases') & (data.NumValue!='-') & (data.RegionName=='Italy')]
dataIT['Time'] = pd.to_datetime(dataIT['Time'])
dataIT['NumValue'] = pd.to_numeric(dataIT['NumValue'])
dataIT.set_index('Time', inplace=True)
dataIT
```

### Temperatura media mensual en Italia

```python
#Ferrara
data=pd.read_csv("./tiempo/Italy/TG_FERRARA.txt", sep=",")
data['DATE'] = pd.to_datetime(data['DATE'],format='%Y%m%d')
data=data[(data.DATE>='2008-01-01') & (data.Q_TG!=9)]
df=data.iloc[:,[0,2,3]]
dataW = df.set_index('DATE', inplace=True)
df_ferrara=df.resample('MS').mean()

#Roma Ciampino
data=pd.read_csv("./tiempo/Italy/TG_ROMA_CIAMPINO.txt", sep=",")
data['DATE'] = pd.to_datetime(data['DATE'],format='%Y%m%d')
data=data[(data.DATE>='2008-01-01') & (data.Q_TG!=9)]
df=data.iloc[:,[0,2,3]]
dataW = df.set_index('DATE', inplace=True)
df_roma=df.resample('MS').mean()

#Bologna
data=pd.read_csv("./tiempo/Italy/TG_BOLOGNA.txt", sep=",")
data['DATE'] = pd.to_datetime(data['DATE'],format='%Y%m%d')
data=data[(data.DATE>='2008-01-01') & (data.Q_TG!=9)]
df=data.iloc[:,[0,2,3]]
dataW = df.set_index('DATE', inplace=True)
df_bologna=df.resample('MS').mean()

#Milan
data=pd.read_csv("./tiempo/Italy/TG_MILAN.txt", sep=",")
data['DATE'] = pd.to_datetime(data['DATE'],format='%Y%m%d')
data=data[(data.DATE>='2008-01-01') & (data.Q_TG!=9)]
df=data.iloc[:,[0,2,3]]
dataW = df.set_index('DATE', inplace=True)
df_milan=df.resample('MS').mean()

#Brindisi
data=pd.read_csv("./tiempo/Italy/TG_BRINDISI.txt", sep=",")
data['DATE'] = pd.to_datetime(data['DATE'],format='%Y%m%d')
data=data[(data.DATE>='2008-01-01') & (data.Q_TG!=9)]
df=data.iloc[:,[0,2,3]]
dataW = df.set_index('DATE', inplace=True)
df_brindisi=df.resample('MS').mean()

#Cagliari
data=pd.read_csv("./tiempo/Italy/TG_CAGLIARI.txt", sep=",")
data['DATE'] = pd.to_datetime(data['DATE'],format='%Y%m%d')
data=data[(data.DATE>='2008-01-01') & (data.Q_TG!=9)]
df=data.iloc[:,[0,2,3]]
dataW = df.set_index('DATE', inplace=True)
df_cagliari=df.resample('MS').mean()

#Cagliari
data=pd.read_csv("./tiempo/Italy/TG_VERONA_VILLAFRANCA.txt", sep=",")
data['DATE'] = pd.to_datetime(data['DATE'],format='%Y%m%d')
data=data[(data.DATE>='2008-01-01') & (data.Q_TG!=9)]
df=data.iloc[:,[0,2,3]]
dataW = df.set_index('DATE', inplace=True)
df_verona=df.resample('MS').mean()


df_concat = pd.concat([df_ferrara, df_roma, df_bologna, df_milan, df_brindisi, df_cagliari, df_verona], axis=0)
df_media_TG = df_concat.groupby(level=0).mean()
df_media_TG
```

### Tratamiento de los valores nulos

```python
nilo.Italy=pd.to_numeric(nilo.Italy)
nilo[nilo.index<'2024-01-01'].info()
```

```python
import numpy as np
import pandas as pd

def interpolate_and_extrapolate(series):
    values = series.values.copy()
    n = len(values)
    
    # Índices no nulos
    non_nan_idx = np.where(~np.isnan(values))[0]
    
    if len(non_nan_idx) < 2:
        raise ValueError("Se necesitan al menos dos valores no nulos para interpolar y extrapolar.")
    
    # Interpolación interna 
    values = np.interp(x=np.arange(n), xp=non_nan_idx, fp=values[non_nan_idx])
    
    # Calculamos el primer y último step
    first_step = (values[non_nan_idx[1]] - values[non_nan_idx[0]]) / (non_nan_idx[1] - non_nan_idx[0])
    last_step = (values[non_nan_idx[-1]] - values[non_nan_idx[-2]]) / (non_nan_idx[-1] - non_nan_idx[-2])
    
    # Extrapolación a la izquierda (inicio)
    for i in range(0, non_nan_idx[0]):
        values[i] = values[non_nan_idx[0]] - first_step * (non_nan_idx[0] - i)
    
    # Extrapolación a la derecha (final)
    for i in range(non_nan_idx[-1] + 1, n):
        values[i] = values[non_nan_idx[-1]] + last_step * (i - non_nan_idx[-1])
        
    return pd.Series(values, index=series.index)

nilo.Italy=interpolate_and_extrapolate(nilo.Italy)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(13,5))
sns.lineplot(data=nilo,x=nilo.index,y=nilo.Italy)
plt.xticks(rotation=90)
plt.show()
```
